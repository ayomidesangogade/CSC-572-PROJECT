# Phishing URL Detection: Machine Learning Pipeline (PhiUSIIL Dataset)

This notebook mirrors the pipeline from `Phishing_URL_Detection.ipynb`, updated to train and
test on the newer, larger **PhiUSIIL Phishing URL Dataset** in place of the original UCI
Phishing Websites dataset.

**Dataset:** PhiUSIIL Phishing URL Dataset — 235,795 labeled URLs described by 54 engineered
features (lexical, HTML/content-based, and behavioral), versus the original UCI dataset's
11,055 URLs and 30 features.

**Pipeline steps (unchanged from the original notebook):**
1. Load and explore the dataset
2. Exploratory data analysis
3. Train/test split
4. Train and compare multiple classification algorithms
5. Evaluate using accuracy, precision, recall, F1-score, and ROC-AUC
6. Analyze feature importance
7. Save the best model

> **Two things to know before running this on your machine:**
> 1. `SVC(probability=True)` is a kernel SVM, and kernel SVMs scale poorly (roughly O(n²)-O(n³))
>    with the number of training rows. At ~189K training rows (vs. ~8.8K before), this cell can
>    take a long time — anywhere from tens of minutes to several hours depending on your CPU.
>    If it's taking too long, the most common fix is to fit the SVM on a stratified subsample of
>    the training set (e.g. 15,000-20,000 rows) rather than the full set; a commented-out
>    version of this is included in the training cell below.
> 2. This notebook still uses `xgboost.XGBClassifier`, exactly like the original. Make sure
>    `pip install -r requirements.txt` has been run in this environment.

## 1. Imports

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import joblib
import time

from sklearn.model_selection import train_test_split, cross_val_score
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC
from sklearn.calibration import CalibratedClassifierCV
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score, f1_score,
    roc_auc_score, roc_curve, confusion_matrix, classification_report
)
from xgboost import XGBClassifier

sns.set_style("whitegrid")
RANDOM_STATE = 42
```

## 2. Load and Explore the Dataset

The dataset (`PhiUSIIL_Phishing_URL_Dataset.csv`) contains 54 engineered features plus a
`label` column. **Note the label convention is the opposite of the old UCI dataset:**
here, `label = 1` means the URL is **legitimate** and `label = 0` means it is **phishing**.
We'll remap this to the same `1 = phishing, 0 = legitimate` convention used throughout the
original notebook and thesis.

We also drop five identifier/text columns (`FILENAME`, `URL`, `Domain`, `TLD`, `Title`) that
are raw identifiers or free text rather than pre-engineered numeric features — analogous to
how the original 30 UCI features were already numeric and required no such filtering.

```python
df = pd.read_csv("PhiUSIIL_Phishing_URL_Dataset.csv")
print("Shape:", df.shape)
df.head()
```

```python
print("Missing values per column:")
print(df.isnull().sum().sum(), "total missing values")
df.describe().T
```

```python
class_counts = df["label"].value_counts().rename({1: "Legitimate", 0: "Phishing"})
print(class_counts)

plt.figure(figsize=(5, 4))
sns.barplot(x=class_counts.index, y=class_counts.values, hue=class_counts.index, palette="viridis", legend=False)
plt.title("Class Distribution")
plt.ylabel("Number of URLs")
plt.show()
```

## 3. Exploratory Data Analysis

```python
id_cols = ["FILENAME", "URL", "Domain", "TLD", "Title"]
numeric_df = df.drop(columns=id_cols)

corr = numeric_df.corr(numeric_only=True)["label"].drop("label").sort_values()

plt.figure(figsize=(8, 12))
corr.plot(kind="barh", color=(corr > 0).map({True: "seagreen", False: "indianred"}))
plt.title("Feature Correlation with Target (label; 1 = Legitimate)")
plt.xlabel("Correlation coefficient")
plt.tight_layout()
plt.savefig("feature_correlation.png", dpi=300)
plt.show()
```

```python
plt.figure(figsize=(16, 13))
sns.heatmap(numeric_df.corr(numeric_only=True), cmap="coolwarm", center=0, cbar_kws={"shrink": 0.7})
plt.title("Feature Correlation Heatmap")
plt.tight_layout()
plt.show()
```

## 4. Train/Test Split

We map the target to the standard convention `1 = phishing`, `0 = legitimate` (note this is
the reverse of the raw `label` column in this dataset), and drop the identifier/text columns
that aren't part of the numeric feature set.

```python
id_cols = ["FILENAME", "URL", "Domain", "TLD", "Title"]
X = df.drop(columns=id_cols + ["label"])
y = df["label"].map({0: 1, 1: 0})  # 1 = phishing, 0 = legitimate

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=RANDOM_STATE, stratify=y
)
print("Train size:", X_train.shape, " Test size:", X_test.shape)
```

### A necessary addition versus the original notebook: feature scaling

The original UCI dataset had every feature already encoded on the same -1/0/1 scale, so the
original notebook fed raw features straight into every model, including `SVC` and
`LogisticRegression`. PhiUSIIL's features are on very different scales (e.g. `LineOfCode` can
run into the thousands next to 0/1 flags like `IsHTTPS`), and both of those two models are
scale-sensitive. Without scaling, `SVC` here scores well below chance-adjacent performance and
`LogisticRegression` fails to converge. Tree-based models (Decision Tree, Random Forest,
XGBoost) are unaffected by scale, so we only scale the copies fed to LR and SVM.

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

## 5. Train and Compare Multiple Models

As in the original notebook, we train and compare five algorithms:
Logistic Regression, Decision Tree, Random Forest, Support Vector Machine, and XGBoost.
Logistic Regression and SVM use the scaled features from above; the tree-based models use the
raw features, same as the original notebook.

**Runtime note:** with ~189K training rows, `SVC(probability=True)` is the slow step here.
If it's taking too long on your machine, comment out the full-data `.fit()` call and use the
subsampled version shown below it instead.

```python
models = {
    "Logistic Regression": (LogisticRegression(max_iter=1000, random_state=RANDOM_STATE), True, False),
    "Decision Tree": (DecisionTreeClassifier(random_state=RANDOM_STATE), False, False),
    "Random Forest": (RandomForestClassifier(n_estimators=200, random_state=RANDOM_STATE, n_jobs=-1), False, False),
    "SVM": (CalibratedClassifierCV(SVC(), ensemble=False), True, True),
    "XGBoost": (XGBClassifier(eval_metric="logloss", random_state=RANDOM_STATE), False, False),
}
# Each entry is (model, use_scaled_features, use_svm_subsample)

# --- Optional: if SVM is too slow on the full training set, subsample just for that model ---
from sklearn.utils import resample
X_train_svm, y_train_svm = resample(
    X_train_scaled, y_train, n_samples=20000, stratify=y_train, random_state=RANDOM_STATE
)

trained_models = {}
model_uses_scaled = {}
results = []

for name, (model, use_scaled, use_subsample) in models.items():
    t0 = time.time()
    if use_subsample:
        Xtr, ytr = X_train_svm, y_train_svm
    else:
        Xtr = X_train_scaled if use_scaled else X_train
        ytr = y_train
    Xte = X_test_scaled if use_scaled else X_test
    model.fit(Xtr, ytr)
    trained_models[name] = model
    model_uses_scaled[name] = use_scaled

    preds = model.predict(Xte)
    probs = model.predict_proba(Xte)[:, 1]

    results.append({
        "Model": name,
        "Accuracy": accuracy_score(y_test, preds),
        "Precision": precision_score(y_test, preds),
        "Recall": recall_score(y_test, preds),
        "F1-score": f1_score(y_test, preds),
        "ROC-AUC": roc_auc_score(y_test, probs),
    })
    print(f"{name} done in {time.time() - t0:.1f}s" + (" (trained on 20K stratified subsample)" if use_subsample else ""))

results_df = pd.DataFrame(results).sort_values("F1-score", ascending=False).reset_index(drop=True)
results_df
```

```python
results_df.set_index("Model")[["Accuracy", "Precision", "Recall", "F1-score", "ROC-AUC"]].plot(
    kind="bar", figsize=(11, 6)
)
plt.title("Model Comparison")
plt.ylabel("Score")
plt.ylim(0.99, 1.001)  # scores are extremely high on this dataset - zoom in accordingly
plt.xticks(rotation=20)
plt.legend(loc="lower right")
plt.tight_layout()
plt.show()
```

## 6. Detailed Evaluation of the Best Model

```python
best_name = results_df.iloc[0]["Model"]
best_model = trained_models[best_name]
best_uses_scaled = model_uses_scaled[best_name]
X_test_best = X_test_scaled if best_uses_scaled else X_test
print("Best model by F1-score:", best_name)

preds = best_model.predict(X_test_best)
probs = best_model.predict_proba(X_test_best)[:, 1]

print(classification_report(y_test, preds, target_names=["Legitimate", "Phishing"]))
```

```python
cm = confusion_matrix(y_test, preds)
plt.figure(figsize=(5, 4))
sns.heatmap(cm, annot=True, fmt="d", cmap="Blues",
            xticklabels=["Legitimate", "Phishing"], yticklabels=["Legitimate", "Phishing"])
plt.title(f"Confusion Matrix: {best_name}")
plt.ylabel("Actual")
plt.xlabel("Predicted")
plt.show()
```

```python
plt.figure(figsize=(6, 5))
for name, model in trained_models.items():
    Xte_i = X_test_scaled if model_uses_scaled[name] else X_test
    probs_i = model.predict_proba(Xte_i)[:, 1]
    fpr, tpr, _ = roc_curve(y_test, probs_i)
    auc = roc_auc_score(y_test, probs_i)
    plt.plot(fpr, tpr, label=f"{name} (AUC={auc:.4f})")

plt.plot([0, 1], [0, 1], linestyle="--", color="gray")
plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("ROC Curves")
plt.legend(loc="lower right", fontsize=8)
plt.tight_layout()
plt.show()
```

## 7. Feature Importance

Understanding which features drive predictions supports the interpretability objective from
the proposal.

```python
if hasattr(best_model, "feature_importances_"):
    importances = pd.Series(best_model.feature_importances_, index=X.columns).sort_values(ascending=False)
elif hasattr(best_model, "coef_"):
    importances = pd.Series(np.abs(best_model.coef_[0]), index=X.columns).sort_values(ascending=False)
else:
    importances = None

if importances is not None:
    plt.figure(figsize=(8, 9))
    importances.head(15).sort_values().plot(kind="barh", color="teal")
    plt.title(f"Top 15 Feature Importances ({best_name})")
    plt.tight_layout()
    plt.show()
else:
    print(f"{best_name} does not expose feature importances directly.")
```

## 8. Cross-Validation Check

A 5-fold cross-validation confirms the result is not an artifact of a single train/test split.

**Note:** this refits `best_model` five times on the *full* dataset (~236K rows each fit). If
`best_model` turns out to be the SVM, this can be very slow for the same reason noted in
Section 5 — consider skipping this cell or reducing to `cv=3` if it stalls. If the best model
needs scaled features, we wrap it in a `Pipeline` so each fold scales using only its own
training data (avoiding leakage across folds).

```python
from sklearn.pipeline import make_pipeline

if best_name == "SVM":
    X_cv, y_cv = resample(X, y, n_samples=20000, stratify=y, random_state=RANDOM_STATE)
    cv_estimator = make_pipeline(StandardScaler(), SVC(probability=True))
else:
    X_cv, y_cv = X, y
    cv_estimator = make_pipeline(StandardScaler(), best_model) if best_uses_scaled else best_model

cv_scores = cross_val_score(cv_estimator, X_cv, y_cv, cv=5, scoring="f1")
print(f"5-fold CV F1-scores: {cv_scores}")
print(f"Mean F1: {cv_scores.mean():.4f}  (+/- {cv_scores.std():.4f})")
```

## 9. Save the Best Model

Saved under new filenames (`_phiusiil` suffix) rather than overwriting `phishing_model.pkl` /
`feature_columns.pkl`, since this model uses a completely different 50-feature schema. **The
existing `app.py` and `feature_extraction.py` are built around the old 30 UCI features and will
not work with this model as-is** — they'd need to be rewritten to compute PhiUSIIL-style
features (e.g. `LineOfCode`, `NoOfJS`, `NoOfImage`, `HasSocialNet`) from a live URL. Happy to
help with that rewrite as a follow-up if you want the demo app updated too.

```python
joblib.dump(best_model, "phishing_model_phiusiil.pkl")
joblib.dump(list(X.columns), "feature_columns_phiusiil.pkl")
print("Saved phishing_model_phiusiil.pkl and feature_columns_phiusiil.pkl")
```
